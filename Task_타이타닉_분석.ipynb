피벗 테이블에서 알 수 있는 사실

- 여성은 모든 객실 등급에서 남성보다 생존율이 높다.
- 여성 중에서는 1등급의 생존율이 가장 높다.
- 남성 중에서도 1등급의 생존율이 가장 높다.
- 여성 1등급의 생존율은 약 96.8%이나 남성 3등급의 생존율은 약 13.5%이다.

```python
#결측치 처리하기
titanic_df.isnull().sum()
```

```python
# 원본데이터 복사하기
titanic_copy = titanic_df.copy()

titanic_copy["Age"] = titanic_copy["Age"].fillna(
    titanic_copy["Age"].mean()
)
print(titanic_df["Age"].isnull().sum())
print(titanic_copy["Age"].isnull().sum())
```

```python
# 1. 평균으로 채우기
age_mean = titanic_df["Age"].fillna(
    titanic_df["Age"].mean()
)

# 2. 중앙값으로 채우기
age_median = titanic_df["Age"].fillna(
    titanic_df["Age"].median()
)

# 3. 결측치가 있는 행 삭제하기
age_drop = titanic_df["Age"].dropna()

print("평균:", age_mean.mean())
print("중앙값:", age_median.mean())
print("삭제 후 데이터 개수:", len(age_drop))
```
